```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
dataset=pd.read_csv("Teen_Mental_Health_Dataset.csv") 
```

```python
dataset
```

```python
dataset.dropna(inplace=True)
```

```python
dataset = pd.get_dummies(dataset, columns=["gender", "platform_usage"], drop_first=True)
```

```python
le = LabelEncoder()
```

```python
dataset["social_interaction_level"] = le.fit_transform(dataset["social_interaction_level"])
```

```python
dataset
```

```python
X = dataset.drop(columns=["social_interaction_level"])
y = dataset["social_interaction_level"]
```
