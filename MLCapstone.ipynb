```python
import pandas as pd
```

```python
dataset=pd.read_csv("Teen_Mental_Health_Dataset.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
```

```python
dataset
```

```python
dataset.isnull().sum()
```

```python
dataset.dropna(inplace=True)
```

```python
dataset
```

```python
dataset["social_interaction_level_medium"].fillna(dataset["social_interaction_level_medium"].mean(),inplace=True)
```

```python
dataset
```

```python
def quanQual(dataset):
    quan=[]
    qual=[]
    for columnName in dataset.columns:
        #print(columnName)
        if(dataset[columnName].dtype=='O'):
            #print("qual")
            qual.append(columnName)
        else:
            #print("quan")
            quan.append(columnName)
    return quan,qual
```

```python
quan,qaul=quanQual(dataset)
```

```python
dataset[quan]
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values=np.nan, strategy='mean')
imp.fit(dataset[quan])
df=imp.transform(dataset[quan])
df = pd.DataFrame(df,columns = quan)
```

```python
df
```

```python
cate=dataset[qaul]
```

```python
cate
```

```python
two=[df,cate]
```

```python
preprocessed=pd.concat(two,axis=1)
```

```python
preprocessed
```

```python
preprocessed.to_csv("social_interaction_level_medium.csv",index=False)
```
